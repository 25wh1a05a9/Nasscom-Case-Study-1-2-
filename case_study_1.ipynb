```python
# Task 1 = using for loop, sensor readsing are displayed:
battery_data = [
    [0, 32, 400, 20, 95],
    [1, 35, 395, 22, 88],
    [2, 38, 390, 25, 80],
    [3, 40, 385, 28, 70],
    [4, 42, 380, 30, 60],
    [5, 45, 375, 32, 48]
]

# Display all readings
for data in battery_data:
    print(data)
```

```python
# Task 1 = using if condition printing tempearature hike and normal:
for data in battery_data:
    temperature = data[1]

    if temperature > 40:
        print("Time:", data[0], "High Temperature Alert")
    else:
        print("Time:", data[0], "Temperature Normal")
```

```python
# Task 1 = using if condtion to identify charge of the battery:
for data in battery_data:
    charge = data[4]

    if charge < 50:
        print("Time:", data[0], "Low Battery Alert")
    else:
        print("Time:", data[0], "Battery Charge Normal")
```

```python
#Task -1 = creation of function:
def battery_status(charge):
    if charge > 70:
        return "High"
    elif charge >= 50:
        return "Medium"
    else:
        return "Low"
for data in battery_data:
    print("Time:", data[0], "Charge:", data[4],
          "Status:", battery_status(data[4]))
```

```python
#Task 2 = Numpy array creation
import numpy as np

D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])

# 1. Shape
print("Shape:", D.shape)

# 2. Average of each parameter
print("Average Temperature:", np.mean(D[:, 0]))
print("Average Voltage:", np.mean(D[:, 1]))
print("Average Current:", np.mean(D[:, 2]))
print("Average Charge:", np.mean(D[:, 3]))

# 3. Maximum and minimum temperature
print("Maximum Temperature:", np.max(D[:, 0]))
print("Minimum Temperature:", np.min(D[:, 0]))

# 4. Extract voltage column
V = D[:, 1]

print("Voltage Column:")
print(V)

# 5. High-temperature readings
high_temperature = D[D[:, 0] > 40]

print("High Temperature Readings:")
print(high_temperature)

# 6. Low battery readings
low_battery = D[D[:, 3] < 50]

print("Low Battery Readings:")
print(low_battery)
```

```python
#Task-3 :Algebra
import numpy as np

A = np.array([
    [2, 1],
    [1, 3]
])

X = np.array([
    [40],
    [25]
])

# Matrix multiplication
Y = A @ X

print("Y =")
print(Y)

# Transpose
print("Transpose of A:")
print(A.T)

# Determinant
print("Determinant of A:")
print(np.linalg.det(A))

# Inverse
print("Inverse of A:")
print(np.linalg.inv(A))
```

```python
#Using Numpy
A = np.array([
    [2, 1],
    [1, 3]
])

B = np.array([100, 120])

solution = np.linalg.solve(A, B)

print("x =", solution[0])
print("y =", solution[1])
```
