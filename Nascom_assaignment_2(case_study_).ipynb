```python
#F-Caculus
# By Differentiation

def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

print("x\tF(x)\tF'(x)")

for x in range(1, 11):
    print(x, "\t", F(x), "\t", dF(x))

# Point where derivative is zero
x = 5 / 0.4

print("\nF'(x) = 0 at x =", x)
```

```python
# Part G - Integration

import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)
y = L(x)

area = np.trapezoid(y, x)

print("Numerical integral =", area)
```

```python
# Part I - Mini Prediction Model

import numpy as np
import pandas as pd

data = {
    "Student":["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10"],
    "Study_Hours":[2,3,4,5,6,7,8,4,6,9],
    "Attendance":[65,70,75,80,85,90,92,78,88,95],
    "Assignment_Score":[55,60,65,72,78,85,88,70,80,92],
    "Previous_Score":[58,62,68,70,76,82,86,72,79,90],
    "Final_Score":[60,64,70,75,82,88,91,74,84,94]
}
df = pd.DataFrame(data)

# Input variables
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

# Final Score
Y = df["Final_Score"].values

# Add column of 1s for intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate coefficients
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
